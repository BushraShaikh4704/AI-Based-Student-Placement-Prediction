## Design Database Table

The database structure is designed based on the actual project features and application requirements.  
Student input features, engineered features, placement prediction results, and career recommendation results are identified separately before creating the MySQL table.

## Create Table

In this section, the MySQL table is created using the database design defined in the previous section.  
The table stores student information, input features, engineered features, placement prediction results, and career recommendation results.  
A unique student ID is used as the primary key, and the prediction date and time are stored for each record.

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS students (
    student_id INT AUTO_INCREMENT PRIMARY KEY,
    student_name VARCHAR(100),

    branch VARCHAR(50),
    college_tier VARCHAR(50),
    cgpa FLOAT,
    backlogs INT,

    coding_skills FLOAT,
    dsa_score FLOAT,
    aptitude_score FLOAT,
    communication_skills FLOAT,
    ml_knowledge FLOAT,
    system_design FLOAT,

    internships INT,
    projects_count INT,
    certifications INT,
    hackathons INT,
    open_source_contributions INT,
    extracurriculars INT,

    technical_skill_score FLOAT,
    has_backlog INT,
    experience_score INT,
    technical_skill_gap FLOAT,

    placement_status INT,
    placement_probability FLOAT,

    recommended_career VARCHAR(100),
    career_suitability_score FLOAT,
    alternative_career VARCHAR(100),

    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
"""

cursor.execute(create_table_query)
connection.commit()

print("Students table created successfully!")
```
